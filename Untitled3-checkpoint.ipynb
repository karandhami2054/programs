```python
paragraph = input("Enter your paragraph")
if paragraph.lower() == paragraph[-1::-1].lower():
   print("Palindrome")
else:
    print("Not palindrome")
```

```python
sentence = input("Enter a sentence: ")

reverse = "".join(reversed(sentence))
if sentence.lower() == reverse.lower():
    print("Sentence is a palindrome")
else:
    print("Sentence is not a palindrome")
print(reverse)
```
